# Credit Card Fraud — Unsupervised Anomaly Detection & Transaction Segmentation

This project detects likely fraudulent transactions **without using the fraud label to train
any model** — the label is used only afterward, to check how well the unsupervised methods
would have caught real fraud. This mirrors a real deployment scenario: a system has to flag a
suspicious transaction before anyone knows the true outcome.

Three unsupervised techniques are applied to the same transaction-level dataset:

- **Anomaly detection** (Isolation Forest, Local Outlier Factor) — flagging individual suspicious transactions, with SHAP explaining *why* each one was flagged
- **Segmentation** (K-Means, DBSCAN) — grouping transactions by behavioral pattern, to surface group-level patterns anomaly detection alone would miss
- **PCA** — visualizing separability early, then compressing the feature space in a principled way

Dataset: transaction-level records with features like `distance_from_home`,
`distance_from_last_transaction`, and `ratio_to_median_purchase_price`, plus a `fraud` label
held out of training and used only for evaluation.

## 4. Feature Engineering

`distance_from_home`, `distance_from_last_transaction`, and `ratio_to_median_purchase_price`
are heavily right-skewed — a small number of transactions are extreme by design, which is
exactly what makes them worth flagging. A `log1p` transform pulls in the long tail so
distance-based models (K-Means, LOF, PCA) aren't dominated by a handful of extreme values.

```python
df['distance_from_home_log'] = np.log1p(df['distance_from_home'])
df['distance_from_last_transaction_log'] = np.log1p(df['distance_from_last_transaction'])
df['ratio_to_median_purchase_price_log'] = np.log1p(df['ratio_to_median_purchase_price'])

log.info('Feature Engineering complete. Shape %s', df.shape)
```

## 5. Preprocessing

Builds a `ColumnTransformer`: numeric features are median-imputed and standard-scaled (required
for any distance-based unsupervised method); categorical features are routed by cardinality —
low-cardinality columns are one-hot encoded, high-cardinality columns get a count encoder,
avoiding a sparse explosion. The 2D PCA projection (Fig. 4) gives a first visual read on how
separable the data looks before any model is fit.

```python
cols_to_drop = cfg.drop_cols.copy()
if cfg.label_col and cfg.label_col in df.columns:
    cols_to_drop.append(cfg.label_col)

X = df.drop(columns=cols_to_drop, errors='ignore')

y_eval = df[cfg.label_col] if (cfg.label_col and cfg.label_col in df.columns) else None

if not cfg.numeric_features and not cfg.categorical_features:
    cfg.categorical_features = X.select_dtypes(include=['object','category']).columns.tolist()
    cfg.numeric_features = X.select_dtypes(include=['number']).columns.tolist()
    log.info('Auto-detected %d numeric | %d categorical', len(cfg.numeric_features), len(cfg.categorical_features))

print(f'Numeric ({len(cfg.numeric_features)}):     {cfg.numeric_features[:8]} ...')
print(f'Categorical ({len(cfg.categorical_features)}): {cfg.categorical_features}')
```

```python
low_card_features, high_card_features = [], []

for col in cfg.categorical_features:
    if X[col].nunique() <= cfg.cardinality_threshold:
        low_card_features.append(col)
    else:
        high_card_features.append(col)

cardinality_report = pd.DataFrame({
    'column': cfg.categorical_features,
    'n_unique': [X[c].nunique() for c in cfg.categorical_features],
    'encoder':  ['OneHotEncoder' if X[c].nunique() <= cfg.cardinality_threshold 
                else 'CatBoostEncoder' for c in cfg.categorical_features]
}).sort_values('n_unique', ascending=False) if cfg.categorical_features else pd.DataFrame()

log.info('Low-cardinality  -> OneHot   : %s', low_card_features)
log.info('High-cardinality -> CatBoost : %s', high_card_features)

if not cardinality_report.empty:
    print('\n── Cardinality Routing Report ──')
    display(cardinality_report)
```

```python
numeric_pipe = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

low_card_pipe = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
    ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

high_card_pipe = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
    ('encoder', ce.CountEncoder(normalize=True))
])

transformers = [('num', numeric_pipe, cfg.numeric_features)]
if low_card_features:
    transformers.append(('low_card', low_card_pipe, low_card_features))
if high_card_features:
    transformers.append(('high_card', high_card_pipe, high_card_features))

preprocessor = ColumnTransformer(transformers=transformers)

X_scaled = preprocessor.fit_transform(X)
log.info('Preprocessor fit. X_scaled shape: %s', X_scaled.shape)


if y_eval is not None:
    train_idx, test_idx = train_test_split(
        np.arange(X_scaled.shape[0]),
        test_size=cfg.test_size,
        random_state=cfg.random_state,
        stratify=y_eval
    )
    log.info('Held-out split for anomaly evaluation: %d train / %d test rows.',
             len(train_idx), len(test_idx))
else:
    train_idx = np.arange(X_scaled.shape[0])
    test_idx = None
    log.info('No labels - anomaly models will be fit and scored on all rows (no held-out eval possible).')
```

```python
pca_vis = PCA(n_components=2, random_state=cfg.random_state)
X_2d = pca_vis.fit_transform(X_scaled)

fig, ax = plt.subplots(figsize=(9, 6))

if y_eval is not None:
    scatter = ax.scatter(X_2d[:, 0], X_2d[:,1],
                        c=y_eval.values, cmap='coolwarm',
                        alpha=0.4, s=8, label='Coloured by label')
    plt.colorbar(scatter, ax=ax, label=cfg.label_col)
    ax.set_title('Fig 4 — 2D PCA Projection (coloured by label)', fontweight='bold')
else:
    ax.scatter(X_2d[:, 0], X_2d[:, 1], alpha=0.4, s=8, color='#2196F3')
    ax.set_title('Fig 4 — 2D PCA Projection', fontweight='bold')

ax.set_xlabel(f'PC1 ({pca_vis.explained_variance_ratio_[0]*100:.1f}%)')
ax.set_ylabel(f'PC2 ({pca_vis.explained_variance_ratio_[1]*100:.1f}%)')
plt.tight_layout()
plt.savefig('fig4_pca_projection.png', bbox_inches='tight')
plt.show()

log.info('2D PCA explains %.1f%% of total variance.',
         sum(pca_vis.explained_variance_ratio_) * 100)
```

## 6. Anomaly Detection — Isolation Forest & Local Outlier Factor

Two different approaches are fit and compared side by side:

- **Isolation Forest** — isolates points by random recursive splitting; anomalies take fewer
  splits to isolate.
- **Local Outlier Factor** — flags points whose local neighborhood density is much lower than
  their neighbors', catching anomalies that aren't globally extreme but are locally unusual.

Both are trained with **no access to the fraud label** — `contamination` sets the *expected*
fraud rate, not a target to fit against. The label is used afterward only to check ROC-AUC and
precision on a held-out split, and to run a contamination sensitivity analysis so the alert
threshold can be tuned to a business's actual review capacity instead of picked arbitrarily.
SHAP is applied on top of Isolation Forest to explain *why* an individual transaction was
flagged — the reason an analyst needs before blocking a card.

```python
def build_anomaly_model(name, model, X_fit, X_score, y_score=None):

    log.info('── Anomaly Detection: %s ──', name)

    model.fit(X_fit)

    raw_preds = model.predict(X_score)
    y_pred = (raw_preds == -1).astype(int)

    if hasattr(model, 'score_samples'):
        scores = -model.score_samples(X_score)

    elif hasattr(model, 'negative_outlier_factor_') and X_score is X_fit:
        scores = -model.negative_outlier_factor_
        
    else:
        scores = y_pred.astype(float)

    n_flagged = y_pred.sum()
    log.info('%s | Flagged %d/%d scored rows (%.2f%%)', name, n_flagged, len(y_pred), n_flagged / len(y_pred) * 100)

    res = {'model':model, 'y_pred':y_pred, 'scores':scores, 'n_flagged':n_flagged}
    
        # ── Evaluation (only if labels exist) ────────────────────────────────────
    if y_score is not None:
        if X_score is X_fit:
            log.warning('%s | Evaluating on the same rows used to fit -> metrics below are IN-SAMPLE, not a held-out estimate.', name)

        auc = roc_auc_score(y_score, scores)
        ap  = average_precision_score(y_score, scores)
        log.info('%s | ROC-AUC: %.4f | Avg Precision: %.4f', name, auc, ap)
        print(f'\n{"+"*60}')
        print(f'  {name}')
        print(f'  Scored rows : {len(y_pred)}{"  (held-out test split)" if X_score is not X_fit else "  (IN-SAMPLE - same rows as fit)"}')
        print(f'  Flagged: {n_flagged} ({n_flagged/len(y_pred)*100:.2f}%)')
        print(f'  ROC-AUC        : {auc:.4f}')
        print(f'  Avg Precision  : {ap:.4f}')
        print(classification_report(y_score, y_pred,
                                    target_names=['Normal', 'Anomaly'], digits=3))

        # Confusion matrix
        cm = confusion_matrix(y_score, y_pred)
        fig, ax = plt.subplots(figsize=(5, 4))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax,
                    xticklabels=['Pred: Normal', 'Pred: Anomaly'],
                    yticklabels=['Actual: Normal', 'Actual: Anomaly'])
        ax.set_title(f'Confusion Matrix — {name}', fontweight='bold')
        plt.tight_layout()
        plt.savefig(f'cm_{name.replace(" ", "_")}.png', bbox_inches='tight')
        plt.show()

        res['auc'] = auc
        res['ap']  = ap
    else:
        print(f'\n  {name} | Flagged: {n_flagged} ({n_flagged/len(y_pred)*100:.2f}%) | No labels — skipping AUC.')

    return res
```

```python
# IsolationForest ========

iso_forest = IsolationForest(
    contamination=cfg.contamination,
    n_estimators=200,
    max_samples='auto',
    random_state=cfg.random_state,
    n_jobs=-1
)

X_fit_anom = X_scaled[train_idx]
X_eval_anom = X_scaled[test_idx] if test_idx is not None else X_fit_anom
y_eval_anom = y_eval.iloc[test_idx].reset_index(drop=True) if test_idx is not None else None

results_iso = build_anomaly_model('Isolation Forest', iso_forest, X_fit_anom, X_eval_anom, y_eval_anom)

full_preds_iso = (iso_forest.predict(X_scaled) == -1).astype(int)
full_scores_iso = -iso_forest.score_samples(X_scaled)
```

```python
# ── Local Outlier Factor 

lof = LocalOutlierFactor(
    contamination=cfg.contamination,
    n_neighbors=20,
    algorithm='auto',
    novelty=True,
    n_jobs=-1
)

results_lof = build_anomaly_model('Local Outlier Factor', lof, X_fit_anom, X_eval_anom, y_eval_anom)

full_preds_lof = (lof.predict(X_scaled) == -1).astype(int)
full_scores_lof = -lof.score_samples(X_scaled)
```

```python
# Anomaly model comparison

agreement = (full_preds_iso == full_preds_lof).mean()
both_flag = ((full_preds_iso == 1) & (full_preds_lof == 1)).sum()

print(f'Agreement between IF and LOF : {agreement*100:.1f}%')
print(f'Records flagged by BOTH models: {both_flag}')
print(f'\n💡 Records flagged by BOTH are highest-confidence anomalies.')
print(f'   Use these for immediate analyst review.')

# Add flags to original dataframe
df['flag_iso'] = full_preds_iso
df['flag_lof'] = full_preds_lof
df['flag_both'] = ((df['flag_iso'] == 1) & (df['flag_lof'] == 1)).astype(int)
df['score_iso'] = full_scores_iso
df['score_lof'] = full_scores_lof

log.info('Flags added to dataframe (scored on 100%% of rows, no labels used).')
```

```python
# ── Fig 5: Anomaly Scores Distribution ──
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

for ax, (name, scores) in zip(axes,[
                             ('Isolation Forest', full_scores_iso),
                             ('Local Outlier Factor', full_scores_lof)]):
    ax.hist(scores, bins=60, color='#2196F3', edgecolor='white', alpha=0.8)
    thresh = np.percentile(scores, (1 - cfg.contamination)* 100)
    ax.axvline(thresh, color='#F44336', linestyle='--',
              label=f'Threshold @ contamination={cfg.contamination}')
    ax.set_title(f'Anomaly Score — {name}', fontweight='bold')
    ax.set_xlabel('Score (higher = more anomalous)')
    ax.set_ylabel('Count')
    ax.legend()

plt.suptitle('Fig 5 — Anomaly Score Distributions (full dataset)', fontsize=13)
plt.tight_layout()
plt.savefig('fig5_anomaly_scores.png', bbox_inches='tight')
plt.show()
```

```python
# ── Contamination Sensitivity Analysis ──

contamination_range = np.arange(0.005, 0.151, 0.005)
sensitivity_rows = []

for c in contamination_range:
    iso_c = IsolationForest(
        contamination=c, n_estimators=100,
        random_state=cfg.random_state, n_jobs=-1
    )
    iso_c.fit(X_fit_anom)
    preds = (iso_c.predict(X_eval_anom) == -1).astype(int)
    row = {'contamination':round(c,3), 'flagged_count':preds.sum(),
          'flagged_pct': round(preds.mean()* 100, 2)}
    if y_eval_anom is not None:
        row['precision'] = round(precision_score(y_eval_anom, preds, zero_division=0), 3)
        row['recall'] = round(recall_score(y_eval_anom, preds, zero_division=0), 3)
    sensitivity_rows.append(row)

sensitivity_df = pd.DataFrame(sensitivity_rows)
print('── Contamination Sensitivity Table (held-out evaluation rows) ──')
display(sensitivity_df)
print(f'\n💡 Current cfg.contamination = {cfg.contamination}')
print('   Choose the row that fits your business review capacity.')
```

```python
# ── Fig 7: SHAP Explainability — WHY was this record flagged? ─────────────────

SHAP_SAMPLE = min(2000, X_scaled.shape[0])
shap_sample_idx = random_sample_index(X_scaled.shape[0], SHAP_SAMPLE, cfg.random_state)

X_shap = pd.DataFrame(
    X_scaled[shap_sample_idx],
    columns=(
        cfg.numeric_features +
        (list(preprocessor.named_transformers_['low_card']
              .named_steps['encoder']
              .get_feature_names_out(low_card_features)) if low_card_features else []) +
        high_card_features
    )
)

explainer  = shap.TreeExplainer(iso_forest)
shap_values = explainer.shap_values(X_shap)
log.info('SHAP computed on %d randomly sampled rows.', SHAP_SAMPLE)
```

```python
# ── Fig 7a: Global SHAP — which features drive anomaly scores? ────────────────
fig = plt.figure(figsize=(10, 6))
shap.summary_plot(shap_values, X_shap, plot_type='bar', show=False)
plt.title('Fig 7a — SHAP Global Feature Importance (Isolation Forest)', fontweight='bold')
plt.tight_layout()
plt.savefig('fig7a_shap_global.png', bbox_inches='tight')
plt.show()
```

```python
# ── Fig 7b: Single record waterfall — explain one flagged case ──

flagged_mask_sample = full_preds_iso[shap_sample_idx] == 1
flagged_indices = np.where(flagged_mask_sample)[0]

if len(flagged_indices) > 0:
    record_idx = flagged_indices[0]

    expected = explainer.expected_value
    if hasattr(expected, '__len__'):
        expected = float(expected[0])
    else:
        expected = float(expected)

    exp = shap.Explanation(
        values       = shap_values[record_idx].astype(float),
        base_values  = np.float64(expected),
        data         = np.array(X_shap.iloc[record_idx], dtype=float),
        feature_names= X_shap.columns.tolist()
    )
    shap.waterfall_plot(exp, show=False)
    plt.title(f'Fig 7b — SHAP Waterfall: Record #{record_idx} (Flagged Anomaly)', fontweight='bold')
    plt.tight_layout()
    plt.savefig('fig7b_shap_waterfall.png', bbox_inches='tight')
    plt.show()
else:
    print('ℹ️  No flagged anomalies in the SHAP sample to explain.')
```

## 7. Unsupervised Segmentation of Transaction Behavior

**Framing note:** this section clusters individual *transactions*, not customers — the dataset
has no customer identifier, so this is transaction-level behavioral segmentation, not customer
segmentation. (A true customer segmentation project needs data aggregated per customer — that's
a separate, later piece of this portfolio, built on its own dataset.)

The goal here complements anomaly detection: instead of flagging individual outliers, K-Means
and DBSCAN group transactions into behavioral clusters, which can reveal a *pattern* that
individual-level anomaly detection would miss — e.g. an entire cluster sharing a suspicious
combination of traits, even if no single transaction in it is extreme enough to be flagged
alone. `find_best_k()` selects K by silhouette score, with a manual override available in
`Config` if the business has a target number of segments in mind. DBSCAN is included as a
density-based check — unlike K-Means it doesn't force every point into a cluster, so it
naturally separates out transactions that don't fit any dense group.

```python
# find best k

def find_best_k(X_scaled, cfg, k_range=None, cluster_sample_idx=None):

    if k_range is None:
        k_range = range(2, cfg.kmeans_max_k + 1)

    X_sample = X_scaled if cluster_sample_idx is None else X_scaled[cluster_sample_idx]
    
    inertias    = []
    silhouettes = []

    for k in k_range:
        km       = KMeans(n_clusters=k, random_state=cfg.random_state, n_init=10)
        labels_k = km.fit_predict(X_sample)
        inertias.append(km.inertia_)
        silhouettes.append(
            silhouette_score(X_sample, labels_k,
                             sample_size=3000, random_state=cfg.random_state)
        )
        log.info('K=%d | Inertia=%.1f | Silhouette=%.4f', k, km.inertia_, silhouettes[-1])

    # ── Auto-select: K with highest silhouette ────────────────────────────────
    best_k_idx = int(np.argmax(silhouettes))
    best_k     = list(k_range)[best_k_idx]

    # ── Manual override from Config ───────────────────────────────────────────
    if cfg.n_clusters > 0:
        best_k = cfg.n_clusters
        log.info('Manual n_clusters override: K=%d', best_k)
    else:
        log.info('Auto-selected K=%d (best silhouette=%.4f)', best_k, silhouettes[best_k_idx])

    print(f'\n💡 Recommended K: {best_k}  (silhouette={silhouettes[best_k_idx]:.4f})')
    print(f'   Override by setting cfg.n_clusters in Section 1.')

    return best_k, {
        'k_range'    : list(k_range),
        'inertias'   : inertias,
        'silhouettes': silhouettes,
        'best_k'     : best_k,
    }

log.info('find_best_k() defined.')
```

```python
# Run find_best_k

CLUSTER_SAMPLE  = min(10000, X_scaled.shape[0])
cluster_sample_idx = random_sample_index(X_scaled.shape[0], CLUSTER_SAMPLE, cfg.random_state)

best_k, k_scan = find_best_k(X_scaled, cfg, cluster_sample_idx=cluster_sample_idx)
```

## 8. Dimensionality Reduction (PCA)

Section 5 used a 2-component PCA purely for visualization. This section fits a full PCA to
decide, in a principled way, how many components are needed to retain most of the signal
(`cfg.pca_variance_threshold`) — useful for compressing this feature space before feeding it
into a downstream supervised model, or to reduce the curse-of-dimensionality effect on K-Means
and LOF. The loading plot (Fig. 12) shows which original features build each component, so the
components stay interpretable rather than becoming an opaque transformation.

```python
# ── Fig 11: Explained Variance — how many components to keep? ─────────────────
pca_full = PCA(random_state=cfg.random_state)
pca_full.fit(X_scaled)

cumvar = np.cumsum(pca_full.explained_variance_ratio_)
n_auto = int(np.argmax(cumvar >= cfg.pca_variance_threshold)) + 1

if cfg.n_components_pca > 0:
    n_keep = cfg.n_components_pca
    log.info('Manual n_components_pca=%d', n_keep)
else:
    n_keep = n_auto
    log.info('Auto-selected %d components (explains %.0f%% variance)', n_keep, cfg.pca_variance_threshold*100)

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].bar(range(1, len(pca_full.explained_variance_ratio_)+1),
            pca_full.explained_variance_ratio_, color='#90CAF9', edgecolor='white')
axes[0].set_xlabel('Component')
axes[0].set_ylabel('Explained Variance Ratio')
axes[0].set_title('Per-Component Variance', fontweight='bold')
axes[0].set_xlim(0, min(30, X_scaled.shape[1]) + 1)

axes[1].plot(range(1, len(cumvar)+1), cumvar, 'o-', color='#2196F3', lw=2)
axes[1].axhline(cfg.pca_variance_threshold, color='gray', linestyle='--',
                label=f'{cfg.pca_variance_threshold*100:.0f}% threshold')
axes[1].axvline(n_keep, color='#F44336', linestyle='--', label=f'Keep {n_keep} components')
axes[1].set_xlabel('Number of Components')
axes[1].set_ylabel('Cumulative Explained Variance')
axes[1].set_title('Cumulative Variance (Scree Plot)', fontweight='bold')
axes[1].legend()
axes[1].set_xlim(0, min(30, X_scaled.shape[1]) + 1)

plt.suptitle('Fig 11 — PCA: How Many Components to Keep?', fontsize=13)
plt.tight_layout()
plt.savefig('fig11_pca_variance.png', bbox_inches='tight')
plt.show()

print(f'\n💡 {n_keep} components explain {cumvar[n_keep-1]*100:.1f}% of total variance.')
print(f'   Set cfg.n_components_pca={n_keep} to use this number.')
```

```python
# -- fit final PCA with selected components
pca_final = PCA(n_components=n_keep, random_state=cfg.random_state)
X_pca = pca_final.fit_transform(X_scaled)
pca_cols = [f'PC{i+1}' for i in range(n_keep)]
df_pca = pd.DataFrame(X_pca, columns=pca_cols)

log.info('PCA complete: %d -> %d dimensions | %.1f%% variance retained',
         X_scaled.shape[1], n_keep, cumvar[n_keep-1]*100)
print(f'PCA output shape: {X_pca.shape}')
print(f'\n💡 X_pca (shape {X_pca.shape}) is ready to use as input to:')
print('   - Supervised models (copy-paste into classification_template.ipynb)')
print('   - Clustering (swap X_scaled for X_pca in Section 7)')
print('   - Anomaly detection (swap X_scaled for X_pca in Section 6)')
```

```python
# ── Fig 12: PCA Loading Plot — which original features build each component? ──

loadings_df = pd.DataFrame(
    pca_final.components_.T,
    index=(
        cfg.numeric_features +
        (list(preprocessor.named_transformers_['low_card']
             .named_steps['encoder']
             .get_feature_names_out(low_card_features)) if low_card_features else[]) +
        high_card_features
    ),
    columns=pca_cols
)

# Show top 3 components or all if ≤ 3
components_to_plot = min(3, n_keep)
fig, axes = plt.subplots(1, components_to_plot, figsize=(5 * components_to_plot, 6))
if components_to_plot == 1:
    axes = [axes]

for i, ax in enumerate(axes):
    pc_name = pca_cols[i]
    loading = loadings_df[pc_name].abs().sort_values(ascending=False).head(10)
    colors_l = ['#F44336' if loadings_df[pc_name][f] > 0 else '#2196F3'
                for f in loading.index]
    ax.barh(loading.index, loading.values, color=colors_l, edgecolor='white')
    ax.set_title(f'{pc_name} ({pca_final.explained_variance_ratio_[i]*100:.1f}%)\n(red=positive, blue=negative)',
                 fontweight='bold', fontsize=10)
    ax.set_xlabel('|Loading|')
    ax.invert_yaxis()

plt.suptitle('Fig 12 — PCA Loadings: Top 10 Features per Component', fontsize=12)
plt.tight_layout()
plt.savefig('fig12_pca_loadings.png', bbox_inches='tight')
plt.show()
```

fitted preprocessor and each fitted model are saved separately, alongside the `Config` used to
produce them, so a new batch of transactions can be scored end-to-end by loading this one file —
no retraining required.

```python
# ── Serialize ─────────────────────────────────────────────────────────────────
# 💡 Unlike supervised pipelines, unsupervised models don't wrap the preprocessor
# because fit() doesn't take y. We save preprocessor + model separately.

output = {
    'preprocessor':     preprocessor,    # fitted ColumnTransformer
    'isolation_forest': iso_forest,      # fitted IF
    'lof':              lof,             # fitted LOF
    'kmeans':           kmeans,          # fitted K-Means
    'pca':              pca_final,       # fitted PCA
    'config':           cfg              # save config for reproducibility
}

joblib.dump(output, cfg.model_output)
log.info('All models saved to: %s', cfg.model_output)
print(f'✅ Saved: {cfg.model_output}')

# To reload and score new data:
# saved   = joblib.load(cfg.model_output)
# X_new_s = saved['preprocessor'].transform(X_new)
# anomaly_flags = (saved['isolation_forest'].predict(X_new_s) == -1).astype(int)
# cluster_ids   = saved['kmeans'].predict(X_new_s)
# pca_coords    = saved['pca'].transform(X_new_s)
```

## 10. Summary

This notebook builds an unsupervised fraud-detection pipeline end-to-end: Isolation Forest and
LOF flag individual suspicious transactions without ever training on the fraud label, SHAP
explains *why* a transaction was flagged, and a contamination sensitivity analysis ties the
alert threshold to review capacity rather than an arbitrary cutoff. K-Means and DBSCAN add a
complementary view — behavioral transaction segments — that can surface patterns individual-level
anomaly detection would miss. PCA is used both to visualize separability early on and,
separately, to compress the feature space for any downstream model.

The recurring design choice throughout: every model is trained without the label, and the label
is used only afterward to validate — because in production, a transaction is scored before
anyone knows the truth.
